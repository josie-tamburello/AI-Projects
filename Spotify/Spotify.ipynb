## Data Loading

In this section, we load the dataset and display the first few rows to ensure it's loaded correctly.


```python
import pandas as pd

df = pd.read_csv('spotifytoptracks.csv')

df.head()
```

## Data Cleaning

In this section, we'll clean the dataset by removing unnecessary columns, handling missing values, and eliminating duplicates.


### Handling Missing Values

In this step, we check for missing values in the dataset and handle them by either removing rows with missing values or filling them with appropriate values.


```python
missing_values = df.isnull().sum()
print("Missing values in the dataset:\n", missing_values)

df.dropna(inplace=True)

missing_values_after = df.isnull().sum()
print("Missing values after handling:\n", missing_values_after)
```

### Results and Explanation

From the output, we see that there are **no missing values** in any of the columns. This means that the dataset is already clean with respect to missing data, and no further action was necessary for this step. 

Since we did not find any missing values, the second option (filling missing values) was not needed.



### Removing Duplicate Samples and Features

We now remove any duplicate rows (samples) and check for duplicate columns (features) in the dataset. Duplicate features will be removed if found.


```python
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

df.drop_duplicates(inplace=True)

df = df.loc[:, ~df.columns.duplicated()]

print("Shape of the dataset after cleaning and removing duplicates:", df.shape)
```

### Results and Explanation

#### The Results We Got:
- The shape of the dataset before cleaning was `(50, 17)` (50 rows and 17 columns).
- After removing the unnecessary `Unnamed: 0` column and checking for duplicates, the shape of the dataset is now `(50, 16)` — meaning no duplicate rows or columns were found, but the redundant index column was removed.

#### What These Results Mean:
This indicates that the dataset initially contained a redundant column (`Unnamed: 0`), which was safely removed during the cleaning process. Additionally, no duplicate rows or columns were present, confirming that each track in the dataset is unique. This ensures the integrity of our analysis, as we are working with a clean and accurate dataset without over-representation or redundant features.



### Treating Outliers

Outliers in the data can significantly affect the analysis. We will use the Interquartile Range (IQR) method to detect and remove outliers from the dataset.


```python
numeric_df = df.select_dtypes(include=[np.number])

Q1 = numeric_df.quantile(0.25)
Q3 = numeric_df.quantile(0.75)
IQR = Q3 - Q1

df_cleaned = df[~((numeric_df < (Q1 - 1.5 * IQR)) | (numeric_df > (Q3 + 1.5 * IQR))).any(axis=1)]

print("Shape of the dataset after treating outliers:", df_cleaned.shape)
df_cleaned.info()
```

### Results and Explanation

#### The Results We Got:
- Before removing outliers, the dataset had **50 rows and 17 columns**.
- After treating the outliers, the dataset now contains **25 rows and 17 columns**. This means that 25 tracks were identified as outliers and removed.
  
#### What These Results Mean:
- The removal of outliers ensures that the analysis focuses on tracks that represent typical values for features like `danceability`, `energy`, and `loudness`, without being skewed by extreme data points. This helps us draw more accurate and reliable conclusions.
  
- By focusing on the core data, we can better identify patterns and trends that are true to the majority of the tracks in the Top 50.

After handling missing values, removing duplicates, and treating outliers, we display the structure of the cleaned dataset.

### Which features are strongly positively correlated? 
### Which features are strongly negatively correlated? 
### Which features are not correlated?

In this section, we perform correlation analysis to find positive, negative, and weak correlations between the features.


```python
numeric_df_cleaned = df_cleaned.select_dtypes(include=[np.number])

correlation_matrix = numeric_df_cleaned.corr()

strong_positive_corr = correlation_matrix[correlation_matrix > 0.5].stack().index.tolist()
strong_negative_corr = correlation_matrix[correlation_matrix < -0.5].stack().index.tolist()
no_correlation = correlation_matrix[(correlation_matrix > -0.1) & (correlation_matrix < 0.1)].stack().index.tolist()

print(f'Strongly positive correlations: {strong_positive_corr}')
print(f'Strongly negative correlations: {strong_negative_corr}')
print(f'No significant correlations: {no_correlation}')
```

**What These Results Mean**:
- There are several strong positive correlations, most notably between `energy` and `loudness`, suggesting that higher energy tracks tend to also be louder.
- A strong negative correlation was found between `danceability` and `liveness`, meaning tracks that are highly danceable tend to be less "live-sounding."
- There were some features that showed no significant correlations, such as `energy` and `danceability`, meaning no clear relationship exists between these attributes in the dataset.

### How does the danceability score compare between Pop, Hip-Hop/Rap, Dance/Electronic, and Alternative/Indie genres? 
### How does the loudness score compare between these genres? 
### How does the acousticness score compare between these genres?

We will calculate the average danceability, loudness, and acousticness scores for Pop, Hip-Hop/Rap, Dance/Electronic, and Alternative/Indie genres.


```python
genres_to_compare = ['Pop', 'Hip-Hop/Rap', 'Dance/Electronic', 'Alternative/Indie']
genre_comparison = df[df['genre'].isin(genres_to_compare)].groupby('genre')[['danceability', 'loudness', 'acousticness']].mean()

print(genre_comparison)
```

**What These Results Mean**:
- Hip-Hop/Rap has the highest average danceability score, making it the most rhythmically engaging genre in the Top 50.
- Dance/Electronic and Hip-Hop/Rap have similar loudness levels, but Pop is slightly quieter, and Alternative/Indie is the quietest among the genres analyzed.
- Acousticness is highest for Alternative/Indie, indicating more natural, unprocessed sound elements, while Dance/Electronic has the lowest acousticness, reflecting the heavy use of synthetic sounds.
